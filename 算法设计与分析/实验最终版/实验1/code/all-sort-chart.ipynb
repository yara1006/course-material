```python
import numpy as np
import matplotlib.pyplot as plt
# O（n^2）复杂度算法
# 输入数据
nn_x = np.array([10000, 20000, 30000, 40000, 50000])
select_sort = np.array([0.0841, 0.31955, 0.7147, 1.2885, 2.0095])
bubble_sort = np.array([0.1824, 0.7094, 1.6319, 2.94165, 4.77785])
insert_sort = np.array([0.0639, 0.23805, 0.5296, 0.9652, 1.4967])

# 使用第一个数据点计算比例因子
select_factor = select_sort[0] / (nn_x[0] ** 2)  # 选择排序的比例因子
bubble_factor = bubble_sort[0] / (nn_x[0] ** 2)  # 冒泡排序的比例因子
insert_factor = insert_sort[0] / (nn_x[0] ** 2)  # 插入排序的比例因子

# 存储预测值
select_sort_predicted = (nn_x ** 2) * select_factor
bubble_sort_predicted = (nn_x ** 2) * bubble_factor
insert_sort_predicted = (nn_x ** 2) * insert_factor

# 计算百分比误差
select_sort_percentage_error = ((select_sort_predicted - select_sort) / select_sort) * 100
bubble_sort_percentage_error = ((bubble_sort_predicted - bubble_sort) / bubble_sort) * 100
insert_sort_percentage_error = ((insert_sort_predicted - insert_sort) / insert_sort) * 100

# 打印百分比误差
print("Select Sort Percentage Errors:", select_sort_percentage_error)
print("Bubble Sort Percentage Errors:", bubble_sort_percentage_error)
print("Insert Sort Percentage Errors:", insert_sort_percentage_error)

# 绘制选择排序
plt.plot(nn_x, select_sort, 'o:', label='select sort')
plt.plot(nn_x, select_sort_predicted,'r-o' ,label='select sort predicted')
plt.legend(loc='upper left')
plt.xlabel("scale of n")
plt.ylabel("time[s]")
plt.title("Selection Sort: Actual vs Predicted")
plt.show()

# 绘制冒泡排序
plt.plot(nn_x, bubble_sort, 'o:', label='bubble sort')
plt.plot(nn_x, bubble_sort_predicted, 'r-o' ,label='bubble sort predicted')
plt.legend(loc='upper left')
plt.xlabel("scale of n")
plt.ylabel("time[s]")
plt.title("Bubble Sort: Actual vs Predicted")
plt.show()

# 绘制插入排序
plt.plot(nn_x, insert_sort, 'o:', label='insert sort')
plt.plot(nn_x, insert_sort_predicted, 'r-o' ,label='insert sort predicted')
plt.legend(loc='upper left')
plt.xlabel("scale of n")
plt.ylabel("time[s]")
plt.title("Insertion Sort: Actual vs Predicted")
plt.show()

# 打印预测值数组
print("Select Sort Predicted:", select_sort_predicted)
print("Bubble Sort Predicted:", bubble_sort_predicted)
print("Insert Sort Predicted:", insert_sort_predicted)
```

```python
# o(n log n)
import numpy as np
import matplotlib.pyplot as plt

# 数据
nn_x = np.array([10000, 20000, 30000, 40000, 50000])
merge_sort = np.array([0.0154, 0.0325, 0.0529, 0.0726, 0.0922])
quick_sort = np.array([0.0114, 0.0253, 0.0398, 0.0545, 0.0699])

# 存储归并排序、快速排序的预测值
merge_sort_predicted = []
quick_sort_predicted = []

# 以第一个元素作为预测值基准
merge_sort_base = merge_sort[0]
quick_sort_base = quick_sort[0]

# 计算归并排序和快速排序的预测值
for n in nn_x:
    # 归并排序的预测值，基于第一个元素
    merge_sort_predicted_value = n * np.log(n) * merge_sort_base / (10000 * np.log(10000))
    merge_sort_predicted.append(round(merge_sort_predicted_value, 4))  # 保留4位小数
    
    # 快速排序的预测值，基于第一个元素
    quick_sort_predicted_value = n * np.log(n) * quick_sort_base / (10000 * np.log(10000))
    quick_sort_predicted.append(round(quick_sort_predicted_value, 4))  # 保留4位小数

# 绘制归并排序图
plt.plot(nn_x, merge_sort, 'o:',label='Merge Sort')
plt.plot(nn_x, merge_sort_predicted, 'r-o', label='Merge Sort Predicted')

plt.legend(loc='upper left')
plt.xlabel("Scale of n")
plt.ylabel("Time [s]")
plt.title("Merge Sort: Actual vs Predicted")
plt.show()

# 绘制快速排序图
plt.plot(nn_x, quick_sort,'o:',label='Quick Sort')
plt.plot(nn_x, quick_sort_predicted, 'r-o', label='Quick Sort Predicted')

plt.legend(loc='upper left')
plt.xlabel("Scale of n")
plt.ylabel("Time [s]")
plt.title("Quick Sort: Actual vs Predicted")
plt.show()

# 打印预测值数组
print("Merge Sort Predicted:", merge_sort_predicted)
print("Quick Sort Predicted:", quick_sort_predicted)
```
